## 3. Webcam/CCTV Detection

### i) Webcam/CCTV Detection (grey-scale-image)

```python
front_face_detector = cv2.CascadeClassifier("haarcascade_frontalface_alt2.xml")

video = cv2.VideoCapture(0)

while True:
    success, frame = video.read()
    if success == True:
        grey_image = cv2.cvtColor(frame,cv2.COLOR_BGR2GRAY)
        face_detector = front_face_detector.detectMultiScale(grey_image)
        
        for x,y,w,h in face_detector:
            cv2.rectangle(grey_image,(x,y), (x+w,y+h), (255,0,0),3)
        
        cv2.imshow("image_window",grey_image)
        key = cv2.waitKey(1)
        if key == 81 or key == 113:
           break

video.release()
cv2.destroyAllWindows()
```

### ii) Webcam/CCTV Detection (Colored-image)
